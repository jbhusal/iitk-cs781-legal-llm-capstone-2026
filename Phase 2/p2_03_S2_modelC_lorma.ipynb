# Phase 2 (L-NER) - S2: Model C + LoRMA, 5-fold

Model C (prefix-LM, whole text window read in both directions) with **LoRMA** adapters, then a tagging head. Only the small adapters and the head are trained; Model C's own weights stay frozen. LoRMA changes a weight W to (I + (alpha/r) * B * A) * W, and B starts at zero so training begins exactly at your Phase 1 model.

**Before you run**
1. Accelerator: **GPU** (one T4). Attach: the Phase 2 data dataset, `legallm-tokenizers`, and the Model C checkpoint (`C_step13683_tokens448364544`).
2. First run the quick check with `FOLDS_TO_RUN=[0]` (first code cell). The cell after the model definition prints `self-test passed` and the trainable parameter count (for the report). Then set all five folds and **Save & Run All (Commit)**.
3. Download `submission_raw.zip` and `results.pkl` from the Output tab. LoRMA only trains a few million parameters, so if the F1 is still rising at epoch 10, increase `EPOCHS`.

```python
%%writefile p2_model_c.py
"""Model C (prefix-LM) architecture, copied unchanged from the Phase 1 notebook p1_05 (model + KV-cache generation)."""
import math
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.checkpoint import checkpoint as grad_checkpoint

USE_GRAD_CHECKPOINTING = False   # training here is short; no recomputation needed


def decode(ids):                 # only used by generate_batch (not needed for tagging)
    raise NotImplementedError

class RMSNorm(nn.Module):
    def __init__(self, dim, eps=1e-6):
        super().__init__()
        self.eps = eps
        self.weight = nn.Parameter(torch.ones(dim))

    def forward(self, x):
        dtype = x.dtype
        x32 = x.float()
        return (x32 * torch.rsqrt(x32.pow(2).mean(-1, keepdim=True) + self.eps) * self.weight.float()).to(dtype)


def build_rope_cache(seq_len, d_head, theta=10000.0):
    inv_freq = 1.0 / (theta ** (torch.arange(0, d_head, 2).float() / d_head))
    freqs = torch.outer(torch.arange(seq_len).float(), inv_freq)
    return freqs.cos(), freqs.sin()


def apply_rope(x, cos, sin):
    """x [B,H,T,d]; cos/sin [B or 1, T, d/2]. Rotation in fp32."""
    x1, x2 = x[..., 0::2].float(), x[..., 1::2].float()
    cos, sin = cos[:, None].float(), sin[:, None].float()
    return torch.stack([x1 * cos - x2 * sin, x1 * sin + x2 * cos], dim=-1).flatten(-2).to(x.dtype)


def prefix_lm_mask(prefix_len, T, device):
    """Bool [B, 1, T, T]: key k visible to query q iff k <= q or k < prefix_len[b]."""
    idx = torch.arange(T, device=device)
    causal = idx[None, :, None] >= idx[None, None, :]
    in_prefix = idx[None, None, :] < prefix_len[:, None, None]
    return (causal | in_prefix)[:, None]


class Attention(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        d, h, kv, dh, b = cfg["d_model"], cfg["n_head"], cfg["n_kv_head"], cfg["d_head"], cfg["bias"]
        self.n_head, self.n_kv_head, self.d_head = h, kv, dh
        self.q_proj = nn.Linear(d, h * dh, bias=b)
        self.k_proj = nn.Linear(d, kv * dh, bias=b)
        self.v_proj = nn.Linear(d, kv * dh, bias=b)
        self.o_proj = nn.Linear(h * dh, d, bias=b)
        self.q_norm = RMSNorm(dh) if cfg["qk_norm"] else None
        self.k_norm = RMSNorm(dh) if cfg["qk_norm"] else None

    def forward(self, x, cos, sin, attn_mask, cache=None):
        B, T, _ = x.shape
        q = self.q_proj(x).view(B, T, self.n_head, self.d_head).transpose(1, 2)
        k = self.k_proj(x).view(B, T, self.n_kv_head, self.d_head).transpose(1, 2)
        v = self.v_proj(x).view(B, T, self.n_kv_head, self.d_head).transpose(1, 2)
        if self.q_norm is not None:
            q, k = self.q_norm(q), self.k_norm(k)
        q, k = apply_rope(q, cos, sin), apply_rope(k, cos, sin)
        if cache is not None:
            k, v = cache.update(k, v)
        n_rep = self.n_head // self.n_kv_head
        if n_rep > 1:
            k, v = k.repeat_interleave(n_rep, dim=1), v.repeat_interleave(n_rep, dim=1)
        if attn_mask is None:
            out = F.scaled_dot_product_attention(q, k, v, is_causal=T > 1)
        else:
            out = F.scaled_dot_product_attention(q, k, v, attn_mask=attn_mask)
        return self.o_proj(out.transpose(1, 2).reshape(B, T, self.n_head * self.d_head))


class GeluMLP(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        self.up_proj = nn.Linear(cfg["d_model"], cfg["d_ff"], bias=cfg["bias"])
        self.down_proj = nn.Linear(cfg["d_ff"], cfg["d_model"], bias=cfg["bias"])

    def forward(self, x):
        return self.down_proj(F.gelu(self.up_proj(x), approximate="tanh"))


class Block(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        self.attn_norm = RMSNorm(cfg["d_model"])
        self.attn = Attention(cfg)
        self.mlp_norm = RMSNorm(cfg["d_model"])
        self.mlp = GeluMLP(cfg)

    def _inner(self, x, cos, sin, attn_mask, cache):
        x = x + self.attn(self.attn_norm(x), cos, sin, attn_mask, cache)
        return x + self.mlp(self.mlp_norm(x))

    def forward(self, x, cos, sin, attn_mask, cache=None, use_checkpoint=False):
        if use_checkpoint and cache is None and self.training:
            return grad_checkpoint(self._inner, x, cos, sin, attn_mask, None, use_reentrant=False)
        return self._inner(x, cos, sin, attn_mask, cache)


class KVCache:
    def __init__(self, batch_size, n_kv_head, max_len, d_head, device, dtype):
        self.k = torch.zeros(batch_size, n_kv_head, max_len, d_head, device=device, dtype=dtype)
        self.v = torch.zeros_like(self.k)
        self.pos = 0

    def update(self, k, v):
        T = k.shape[2]
        self.k[:, :, self.pos:self.pos + T] = k
        self.v[:, :, self.pos:self.pos + T] = v
        self.pos += T
        return self.k[:, :, :self.pos], self.v[:, :, :self.pos]


class PrefixLM(nn.Module):
    def __init__(self, config):
        super().__init__()
        self.config = config
        self.embed = nn.Embedding(config["vocab_size"], config["d_model"])
        self.blocks = nn.ModuleList([Block(config) for _ in range(config["n_layer"])])
        self.norm_f = RMSNorm(config["d_model"])
        self.head = nn.Linear(config["d_model"], config["vocab_size"], bias=False)
        self.head.weight = self.embed.weight
        cos, sin = build_rope_cache(config["max_seq_len"], config["d_head"])
        self.register_buffer("rope_cos", cos, persistent=False)
        self.register_buffer("rope_sin", sin, persistent=False)
        self.apply(self._init_weights)
        for name, p in self.named_parameters():
            if name.endswith("o_proj.weight") or name.endswith("down_proj.weight"):
                nn.init.normal_(p, mean=0.0, std=0.02 / math.sqrt(2 * config["n_layer"]))

    @staticmethod
    def _init_weights(module):
        if isinstance(module, (nn.Linear, nn.Embedding)):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)
        if isinstance(module, nn.Linear) and module.bias is not None:
            nn.init.zeros_(module.bias)

    def encode(self, input_ids, attention_mask=None, mode="prefix", prefix_len=None):
        """Contract method: hidden states. mode 'causal' (prefix 0), 'prefix'
        (needs prefix_len [B]) or 'bidirectional' (whole input is prefix)."""
        B, T = input_ids.shape
        if mode == "causal" or prefix_len is None and mode == "prefix":
            prefix_len = torch.zeros(B, dtype=torch.long, device=input_ids.device)
        elif mode == "bidirectional":
            prefix_len = torch.full((B,), T, dtype=torch.long, device=input_ids.device)
        mask = prefix_lm_mask(prefix_len, T, input_ids.device) if bool((prefix_len > 0).any()) else None
        x = self.embed(input_ids)
        cos, sin = self.rope_cos[None, :T], self.rope_sin[None, :T]
        use_ckpt = self.training and USE_GRAD_CHECKPOINTING
        for blk in self.blocks:
            x = blk(x, cos, sin, mask, use_checkpoint=use_ckpt)
        return self.norm_f(x)

    def forward(self, ids, prefix_len=None, targets=None):
        """No targets -> logits. With targets -> [1, 3] = [loss sum, number of
        loss tokens, sum of logsumexp^2 (z-loss)] over positions that predict
        continuation tokens (t >= prefix_len - 1). Sums (not means) so
        DataParallel replicas combine exactly."""
        use_amp = self.training and ids.is_cuda
        if prefix_len is None:
            prefix_len = torch.zeros(ids.shape[0], dtype=torch.long, device=ids.device)
        with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=use_amp):
            logits = self.head(self.encode(ids, mode="prefix", prefix_len=prefix_len))
        if targets is None:
            return logits
        logits32 = logits.float()
        lse = torch.logsumexp(logits32, dim=-1)
        nll = lse - logits32.gather(-1, targets.unsqueeze(-1)).squeeze(-1)
        pos = torch.arange(ids.shape[1], device=ids.device)[None, :]
        w = (pos >= (prefix_len[:, None] - 1)).float()
        return torch.stack([(nll * w).sum(), w.sum(), (lse.pow(2) * w).sum()]).unsqueeze(0)

    def step(self, ids, positions, caches, attn_mask):
        """Generation step with a KV cache. ids/positions [B, T]; attn_mask
        bool [B, 1, T, cache_len + T]."""
        x = self.embed(ids)
        cos, sin = self.rope_cos[positions], self.rope_sin[positions]
        for blk, cache in zip(self.blocks, caches):
            x = blk(x, cos, sin, attn_mask, cache=cache)
        return self.head(self.norm_f(x))

    def generate(self, input_ids, max_new_tokens, kv_cache=True, sampler="greedy", **kw):
        """Contract method (greedy, one prompt): the prompt is the prefix."""
        assert kv_cache and sampler == "greedy"
        toks = generate_batch(self, [input_ids[0].tolist()], max_new_tokens=max_new_tokens, **kw)[0]
        return torch.tensor([input_ids[0].tolist() + toks], device=input_ids.device)


@torch.no_grad()
def generate_batch(model, prompt_id_lists, max_new_tokens=288, max_context=None, rep_penalty=1.0,
                   no_repeat_ngram=0, target_words=None, banned=(), vocab_limit=None):
    """Greedy continuations for several prompts at once (left-padded). The
    prompt is the bidirectional prefix; continuation is causal. Optional:
    repetition penalty over generated tokens, n-gram blocking inside the
    continuation, stop once every row has > target_words words."""
    model.eval()
    device = next(model.parameters()).device
    dtype = next(model.parameters()).dtype
    cfg = model.config
    max_context = max_context or cfg["max_seq_len"]
    prompts = [p[-(max_context - max_new_tokens):] for p in prompt_id_lists]
    B, T0 = len(prompts), max(len(p) for p in prompts)
    pad = torch.tensor([T0 - len(p) for p in prompts], device=device)
    ids = torch.zeros(B, T0, dtype=torch.long, device=device)
    for b, p in enumerate(prompts):
        ids[b, T0 - len(p):] = torch.tensor(p, device=device)
    total = T0 + max_new_tokens
    caches = [KVCache(B, cfg["n_kv_head"], total, cfg["d_head"], device, dtype) for _ in range(cfg["n_layer"])]
    idx = torch.arange(total, device=device)
    key_ok = idx[None, :] >= pad[:, None]
    positions = (idx[None, :] - pad[:, None]).clamp(min=0)
    # Prefill: the prompt attends to itself in both directions (no padding keys);
    # every query may see itself so padded rows are never fully masked.
    mask = (key_ok[:, None, :T0] | torch.eye(T0, dtype=torch.bool, device=device)[None])[:, None]
    logits = model.step(ids, positions[:, :T0], caches, mask)[:, -1, :].float()
    generated = [[] for _ in range(B)]
    seen = [dict() for _ in range(B)]
    for t in range(max_new_tokens):
        if banned:
            logits[:, list(banned)] = -float("inf")
        if vocab_limit is not None:
            logits[:, vocab_limit:] = -float("inf")     # ids the tokenizer cannot decode
        if rep_penalty != 1.0 and t > 0:
            prev = torch.tensor(generated, device=device)
            sc = logits.gather(1, prev)
            logits.scatter_(1, prev, torch.where(sc > 0, sc / rep_penalty, sc * rep_penalty))
        if no_repeat_ngram and t >= no_repeat_ngram - 1:
            n = no_repeat_ngram
            for b in range(B):
                bad = seen[b].get(tuple(generated[b][-(n - 1):]) if n > 1 else ())
                if bad:
                    logits[b, list(bad)] = -float("inf")
        nxt = logits.argmax(dim=-1)
        for b, tok in enumerate(nxt.tolist()):
            generated[b].append(tok)
            if no_repeat_ngram and len(generated[b]) >= no_repeat_ngram:
                g = generated[b][-no_repeat_ngram:]
                seen[b].setdefault(tuple(g[:-1]), set()).add(g[-1])
        if t == max_new_tokens - 1:
            break
        if target_words and t >= 120 and t % 16 == 0 and all(
                len(decode(g).split()) > target_words for g in generated):
            break
        cur = T0 + t
        logits = model.step(nxt[:, None], positions[:, cur:cur + 1], caches,
                            key_ok[:, None, None, :cur + 1])[:, -1, :].float()
    return generated


def build_model(config):
    """Contract function."""
    return PrefixLM(config)
```

```python
# ---- Data, word splitter, tags, Viterbi decoding, strict F1 (same logic as the step-1 offset test) ----
def norm_spans(g):
    out = []
    for x in g:
        if isinstance(x, dict):
            s = x.get("start", x.get("start_char")); e = x.get("end", x.get("end_char")); l = x.get("label", x.get("type"))
        else: s, e, l = x[0], x[1], x[2]
        out.append((int(s), int(e), str(l)))
    return out
def load_fold(fold):
    txt = json.load(open(DATA_DIR / f"fold{fold}-text.json", encoding="utf-8"))
    gold = json.load(open(DATA_DIR / f"fold{fold}-gold.json", encoding="utf-8"))
    if isinstance(txt, list): txt = {str(d.get("id")): d.get("text") for d in txt}
    if isinstance(gold, list): gold = {str(d.get("id")): d.get("entities", d.get("spans", [])) for d in gold}
    return {str(k): v for k, v in txt.items()}, {str(k): norm_spans(v) for k, v in gold.items()}
texts, golds = {}, {}
for f in (1, 2):
    t, g = load_fold(f); texts.update(t); golds.update(g)
test_texts = json.load(open(DATA_DIR / "fold3-text.json", encoding="utf-8"))
if isinstance(test_texts, list): test_texts = {str(d.get("id")): d.get("text") for d in test_texts}
test_texts = {str(k): v for k, v in test_texts.items()}
LABELS = [l.strip() for l in open(DATA_DIR / "labels.txt", encoding="utf-8") if l.strip()]
TAGS = ["O"] + [f"{p}-{l}" for l in LABELS for p in "BI"]
T2I = {t: i for i, t in enumerate(TAGS)}; NT = len(TAGS)
ALL_IDS = sorted(texts)
random.seed(0); _sh = ALL_IDS[:]; random.shuffle(_sh)            # identical to step 1 -> identical folds
K = 5; FOLDS = [sorted(_sh[k::K]) for k in range(K)]
print(f"train docs {len(ALL_IDS)}, test docs {len(test_texts)}, labels {len(LABELS)}, folds {[len(f) for f in FOLDS]}")
print("fold first ids:", [f[:2] for f in FOLDS])

PRE = re.compile(r"[^\W\d_]+|\d+|_|[^\w\s]", re.UNICODE)
def pretok(text): return [(m.start(), m.end()) for m in PRE.finditer(text)]

def spans_to_tags(toks, spans):
    tags = [0] * len(toks)
    starts = {s: i for i, (s, e) in enumerate(toks)}; ends = {e: i for i, (s, e) in enumerate(toks)}
    for s, e, l in sorted(spans):
        if s in starts and e in ends and starts[s] <= ends[e]:
            a, b = starts[s], ends[e]
            if any(t != 0 for t in tags[a:b + 1]): continue
            tags[a] = T2I["B-" + l]
            for k in range(a + 1, b + 1): tags[k] = T2I["I-" + l]
    return tags

def tags_to_spans(toks, path):
    out, cur = [], None
    for (s, e), t in zip(toks, path):
        name = TAGS[t]
        if name.startswith("B-") or (name.startswith("I-") and (cur is None or cur[2] != name[2:])):
            if cur: out.append(tuple(cur))
            cur = [s, e, name[2:]]
        elif name.startswith("I-"): cur[1] = e
        else:
            if cur: out.append(tuple(cur)); cur = None
    if cur: out.append(tuple(cur))
    return out

# allowed tag transitions: I-X may only follow B-X or I-X
TRANS_OK = np.ones((NT, NT), bool)
for j, tj in enumerate(TAGS):
    if tj.startswith("I-"):
        TRANS_OK[:, j] = False
        TRANS_OK[T2I["B-" + tj[2:]], j] = TRANS_OK[j, j] = True
START_OK = np.array([not t.startswith("I-") for t in TAGS])
NEG = -1e9
TRANS_PEN = np.where(TRANS_OK, 0.0, NEG)

def viterbi(logp):
    N = len(logp)
    if N == 0: return []
    score = np.where(START_OK, logp[0], NEG); back = np.zeros((N, NT), np.int16)
    for t in range(1, N):
        cand = score[:, None] + TRANS_PEN
        back[t] = cand.argmax(0); score = cand.max(0) + logp[t]
    path = [int(score.argmax())]
    for t in range(N - 1, 0, -1): path.append(int(back[t][path[-1]]))
    return path[::-1]

def strict_f1(gold, pred, labels=LABELS):
    per = {}
    for l in labels:
        tp = fp = fn = 0
        for d in set(gold) | set(pred):
            g = {x for x in gold.get(d, []) if x[2] == l}; p = {tuple(x[:3]) for x in pred.get(d, []) if x[2] == l}
            tp += len(g & p); fp += len(p - g); fn += len(g - p)
        pr = tp / (tp + fp) if tp + fp else 0.0; rc = tp / (tp + fn) if tp + fn else 0.0
        per[l] = dict(p=pr, r=rc, f1=2 * pr * rc / (pr + rc) if pr + rc else 0.0, support=tp + fn)
    used = [l for l in labels if per[l]["support"] > 0] or labels
    m = lambda k: sum(per[l][k] for l in used) / len(used)
    return m("f1"), m("p"), m("r"), per
```

```python
# ---- Tokenize each word piece with Model D's tokenizer (T2) and cut documents into windows ----
import sentencepiece as spm
sp = spm.SentencePieceProcessor(); sp.load(str(TOK_PATH))
PAD_ID = sp.pad_id() if sp.pad_id() >= 0 else 0
UNK_ID = sp.unk_id()
_cache = {}
def piece_ids(piece, after_space):
    key = (piece, after_space)
    if key in _cache: return _cache[key]
    ids = sp.encode(piece, out_type=int)                 # SentencePiece adds a leading "▁" (word-start marker)
    if not after_space and ids:                          # attached piece (punctuation, footnote digit): drop the marker
        p0 = sp.id_to_piece(ids[0])
        if p0 == "▁": ids = ids[1:]
        elif p0.startswith("▁"):
            alt = sp.piece_to_id(p0[1:])
            if alt != UNK_ID: ids = [alt] + ids[1:]
    if not ids: ids = [UNK_ID]
    _cache[key] = ids
    return ids

class Doc:
    def __init__(self, did, text, spans=None):
        self.id, self.text = did, text
        self.toks = pretok(text)
        ids, first = [], []
        for s, e in self.toks:
            first.append(len(ids)); ids.extend(piece_ids(text[s:e], s == 0 or text[s - 1].isspace()))
        self.ids = np.array(ids, dtype=np.int64); self.first = np.array(first, dtype=np.int64)
        self.n = len(self.ids)
        self.pos2tok = np.full(self.n, -1, dtype=np.int64); self.pos2tok[self.first] = np.arange(len(self.toks))
        self.tags = None
        if spans is not None:
            self.tags = np.array(spans_to_tags(self.toks, spans), dtype=np.int64)
            self.lab = np.full(self.n, -100, dtype=np.int64); self.lab[self.first] = self.tags
            self.gold = spans

def window_starts(n, W, stride, offset=0):
    st = {0, max(0, n - W)} | set(range(offset, max(1, n - W + 1), stride))
    return sorted(st)

t0 = time.time()
DOCS = {i: Doc(i, texts[i], golds[i]) for i in ALL_IDS}
TEST_DOCS = {i: Doc(i, test_texts[i]) for i in sorted(test_texts)}
ntok = sum(d.n for d in DOCS.values()); npiece = sum(len(d.toks) for d in DOCS.values())
print(f"tokenized in {time.time()-t0:.0f}s: {npiece} word pieces -> {ntok} tokens ({ntok/npiece:.2f} tokens/piece); "
      f"test: {sum(d.n for d in TEST_DOCS.values())} tokens")
print("tokens per doc: median", int(np.median([d.n for d in DOCS.values()])), "max", max(d.n for d in DOCS.values()))
print("windows per epoch (all train docs):", sum(len(window_starts(d.n, CFG['WINDOW'], CFG['STRIDE'])) for d in DOCS.values()))
```

```python
# ---- Model C + LoRMA adapters + tagging head; training and prediction code ----
sys.path.insert(0, str(Path.cwd()))
import p2_model_c as legal_c

def _step_of(path): return int(Path(path).stem.split("_step")[1].split("_tokens")[0])
def find_latest_checkpoint(prefix, root, extract_to):
    pts = list(root.rglob(f"{prefix}_step*.pt"))
    if pts: return max(pts, key=_step_of)
    best = None
    for zp in root.rglob("*.zip"):
        try:
            with zipfile.ZipFile(zp) as zf:
                for name in zf.namelist():
                    b = Path(name).name
                    if b.startswith(f"{prefix}_step") and b.endswith(".pt") and (best is None or _step_of(b) > best[0]): best = (_step_of(b), zp, name)
        except zipfile.BadZipFile: continue
    if best is None: return None
    dest = Path(extract_to) / Path(best[2]).name
    with zipfile.ZipFile(best[1]) as zf, zf.open(best[2]) as src, open(dest, "wb") as dst: shutil.copyfileobj(src, dst)
    return dest
CKPT_PATH = Path(os.environ["P2_CKPT"]) if os.environ.get("P2_CKPT") else find_latest_checkpoint("C", KIN, WORK)
assert CKPT_PATH is not None, "No C_step*.pt checkpoint found. Attach the Model C checkpoint dataset."
_ck = torch.load(CKPT_PATH, map_location="cpu", weights_only=False)
C_CONFIG, C_STATE = _ck["config"], _ck["model_state"]
print(f"Model C checkpoint: {CKPT_PATH.name}  step={_ck.get('step')}  tokens={_ck.get('tokens_seen')}")
del _ck

class LoRMALinear(nn.Module):
    """Multiplicative low-rank adaptation:  W' = (I + (alpha/r) B A) W0.   B starts at 0, so W' = W0 at step 0.
    Computed on the layer output:  z = W0 x ;  y = z + (alpha/r) B (A z) + bias."""
    def __init__(self, base, r, alpha):
        super().__init__()
        self.base, self.s = base, alpha / r
        for p in self.base.parameters(): p.requires_grad = False
        d_out = base.out_features
        self.lorma_A = nn.Parameter(torch.randn(r, d_out) / math.sqrt(d_out))
        self.lorma_B = nn.Parameter(torch.zeros(d_out, r))
    def forward(self, x):
        z = F.linear(x, self.base.weight)                                   # W0 x   (no bias yet)
        z = z + self.s * F.linear(F.linear(z, self.lorma_A.to(z.dtype)), self.lorma_B.to(z.dtype))
        return z + self.base.bias.to(z.dtype) if self.base.bias is not None else z
    def merged_weight(self):                                                 # (I + s B A) W0, used only for the self-test
        return self.base.weight + self.s * (self.lorma_B @ (self.lorma_A @ self.base.weight))

TARGETS = ("q_proj", "k_proj", "v_proj", "o_proj", "up_proj", "down_proj")
def add_lorma(model, r, alpha):
    n = 0
    for mod in list(model.modules()):
        for name, child in list(mod.named_children()):
            if name in TARGETS and isinstance(child, nn.Linear):
                setattr(mod, name, LoRMALinear(child, r, alpha)); n += 1
    return n

class NERModel(nn.Module):
    def __init__(self, dropout):
        super().__init__()
        self.base = legal_c.build_model(C_CONFIG); self.base.load_state_dict(C_STATE)
        for p in self.base.parameters(): p.requires_grad = False
        self.n_lorma = add_lorma(self.base, CFG["LORMA_R"], CFG["LORMA_ALPHA"])
        self.drop = nn.Dropout(dropout); self.head = nn.Linear(C_CONFIG["d_model"], NT)
        nn.init.normal_(self.head.weight, std=0.02); nn.init.zeros_(self.head.bias)
    def forward(self, ids, mask):
        lens = mask.sum(1)                                   # the whole real window is the (two-way) prefix; padding stays causal
        with torch.autocast("cuda", dtype=torch.float16, enabled=ids.is_cuda):
            h = self.base.encode(ids, mode="prefix", prefix_len=lens)
        return self.head(self.drop(h.float()))

# ---- self-tests (CPU, a few seconds) ----
def _self_test():
    torch.manual_seed(0)
    tiny = dict(C_CONFIG, d_model=64, n_layer=2, n_head=4, n_kv_head=2, d_head=16, d_ff=128, max_seq_len=64, vocab_size=C_CONFIG["vocab_size"])
    plain = legal_c.build_model(tiny).eval(); wrapped = legal_c.build_model(tiny); wrapped.load_state_dict(plain.state_dict()); wrapped.eval()
    n = add_lorma(wrapped, 4, 8)
    ids = torch.randint(4, 100, (2, 20)); lens = torch.tensor([20, 14])
    with torch.no_grad():
        a = plain.encode(ids, mode="prefix", prefix_len=lens); b = wrapped.encode(ids, mode="prefix", prefix_len=lens)
    d0 = (a - b).abs().max().item(); assert d0 < 1e-5, f"LoRMA changes the model at step 0 ({d0})"
    layer = next(m for m in wrapped.modules() if isinstance(m, LoRMALinear))
    nn.init.normal_(layer.lorma_B, std=0.3); x = torch.randn(3, layer.base.in_features)
    with torch.no_grad():
        ref = F.linear(x, layer.merged_weight(), layer.base.bias); got = layer(x)
    d1 = (ref - got).abs().max().item(); assert d1 < 1e-4, f"adapter != merged (I + sBA)W0 weight ({d1})"
    # tokens inside the real window must not depend on padding
    ids2 = ids.clone(); ids2[1, 14:] = 7
    with torch.no_grad(): c = plain.encode(ids2, mode="prefix", prefix_len=lens)
    d2 = (a[1, :14] - c[1, :14]).abs().max().item(); assert d2 < 1e-4, f"padding leaks into real tokens ({d2})"
    print(f"self-test passed: step-0 identical ({d0:.1e}), adapter = merged weight ({d1:.1e}), padding isolated ({d2:.1e}); {n} layers adapted in the tiny model")
_self_test()
_m = NERModel(0.1)
n_tr = sum(p.numel() for p in _m.parameters() if p.requires_grad); n_all = sum(p.numel() for p in _m.parameters())
print(f"Model C + LoRMA(r={CFG['LORMA_R']}) on {_m.n_lorma} layers: trainable {n_tr/1e6:.2f}M of {n_all/1e6:.1f}M parameters ({100*n_tr/n_all:.1f}%)"); del _m

def batchify(arrs, pad):
    L = max(len(a) for a in arrs); out = np.full((len(arrs), L), pad, dtype=np.int64); m = np.zeros((len(arrs), L), dtype=np.int64)
    for i, a in enumerate(arrs): out[i, :len(a)] = a; m[i, :len(a)] = 1
    return torch.from_numpy(out), torch.from_numpy(m)

@torch.no_grad()
def predict_probs(model, docs, bs=16):
    model.eval(); W, S = CFG["WINDOW"], CFG["STRIDE"]; res = {}
    for d in docs:
        acc = np.zeros((len(d.toks), NT), np.float32); wsum = np.zeros(len(d.toks), np.float32)
        starts = window_starts(d.n, W, S)
        for b in range(0, len(starts), bs):
            chunk = starts[b:b + bs]
            ids, mask = batchify([d.ids[a:a + W] for a in chunk], PAD_ID)
            pr = torch.softmax(model(ids.to(DEVICE), mask.to(DEVICE)), -1).cpu().numpy()
            for k, a in enumerate(chunk):
                L = min(W, d.n - a); ti = d.pos2tok[a:a + L]; keep = ti >= 0
                if not keep.any(): continue
                pos = np.nonzero(keep)[0]; w = 1.0 - np.abs(pos - (L - 1) / 2) / (L / 2 + 1)
                acc[ti[keep]] += pr[k, pos] * w[:, None]; wsum[ti[keep]] += w
        res[d.id] = acc / np.maximum(wsum, 1e-6)[:, None]
    return res

def decode(probs, docs, bias=None):
    out = {}; b = np.zeros(NT, np.float32)
    if bias:
        for l, v in bias.items(): b[T2I["B-" + l]] = b[T2I["I-" + l]] = v
    for d in docs:
        out[d.id] = tags_to_spans(d.toks, viterbi(np.log(np.clip(probs[d.id], 1e-8, 1.0)) + b))
    return out

def dev_score(model, dev_docs):
    pr = predict_probs(model, dev_docs); sp_ = decode(pr, dev_docs)
    return strict_f1({d.id: d.gold for d in dev_docs}, sp_), pr

def tag_weights(train_docs):
    cnt = np.bincount(np.concatenate([d.tags for d in train_docs]), minlength=NT).astype(np.float64) + 1
    w = np.clip((cnt[0] / cnt) ** 0.25, 1.0, 6.0); w[0] = 1.0
    return torch.tensor(w if CFG["CLASS_WEIGHTS"] else np.ones(NT), dtype=torch.float32)

def train_fold(train_docs, dev_docs, fold):
    seed = CFG["SEED"] + fold; random.seed(seed); np.random.seed(seed); torch.manual_seed(seed)
    model = NERModel(CFG["DROPOUT"]).to(DEVICE)
    W, S, BS, AC = CFG["WINDOW"], CFG["STRIDE"], CFG["BATCH"], CFG["ACCUM"]
    n_win = sum(len(window_starts(d.n, W, S)) for d in train_docs)
    total = CFG["EPOCHS"] * math.ceil(n_win / (BS * AC)); warm = max(1, int(CFG["WARMUP"] * total))
    adapters = [p for n, p in model.named_parameters() if p.requires_grad and "lorma_" in n]
    heads = [p for n, p in model.named_parameters() if p.requires_grad and n.startswith("head.")]
    opt = torch.optim.AdamW([dict(params=adapters, lr=CFG["LR_ADAPTER"], weight_decay=0.0), dict(params=heads, lr=CFG["LR_HEAD"], weight_decay=CFG["WD"])])
    base_lrs = [g["lr"] for g in opt.param_groups]; trainable = adapters + heads
    scaler = torch.cuda.amp.GradScaler(enabled=DEVICE.type == "cuda")
    wts = tag_weights(train_docs).to(DEVICE); step = 0; history = []
    print(f"[fold {fold}] train docs {len(train_docs)}, dev docs {len(dev_docs)}, windows/epoch {n_win}, updates {total}")
    for ep in range(CFG["EPOCHS"]):
        model.train(); t0 = time.time(); items = []
        for d in train_docs:
            for a in window_starts(d.n, W, S, offset=random.randrange(S)): items.append((d, a))
        random.shuffle(items); run, nrun = 0.0, 0
        for b in range(0, len(items), BS):
            chunk = items[b:b + BS]
            ids, mask = batchify([d.ids[a:a + W] for d, a in chunk], PAD_ID)
            lab, _ = batchify([d.lab[a:a + W] for d, a in chunk], -100)
            lab = torch.where(mask.bool(), lab, torch.full_like(lab, -100))
            logits = model(ids.to(DEVICE), mask.to(DEVICE))
            loss = F.cross_entropy(logits.view(-1, NT), lab.to(DEVICE).view(-1), weight=wts, ignore_index=-100)
            scaler.scale(loss / AC).backward(); run += loss.item(); nrun += 1
            if (b // BS + 1) % AC == 0 or b + BS >= len(items):
                f = (step + 1) / warm if step < warm else 0.5 * (1 + math.cos(math.pi * min(1.0, (step - warm) / max(1, total - warm))))
                for g, lr0 in zip(opt.param_groups, base_lrs): g["lr"] = lr0 * f
                scaler.unscale_(opt); torch.nn.utils.clip_grad_norm_(trainable, 1.0)
                scaler.step(opt); scaler.update(); opt.zero_grad(set_to_none=True); step += 1
        (f1, pr_, rc_, per), _ = dev_score(model, dev_docs)
        history.append(dict(epoch=ep + 1, loss=run / max(nrun, 1), dev_macro_f1=f1))
        print(f"[fold {fold}] epoch {ep+1}/{CFG['EPOCHS']}  loss {run/max(nrun,1):.4f}  dev strict macro-F1 {f1:.4f} (P {pr_:.3f} R {rc_:.3f})  {time.time()-t0:.0f}s", flush=True)
    return model, history
```

```python
# ---- Run the folds, collect out-of-fold predictions, predict the test set, write submission files ----
oof, test_probs, hist = {}, [], {}
done_path = RUN_DIR / "results.pkl"
if done_path.exists():                                   # resume after a closed tab / restarted session
    _r = pickle.load(open(done_path, "rb")); oof, test_probs, hist = _r["oof"], _r["test_probs"], _r["hist"]
    print("resuming; folds already done:", sorted(hist))
test_docs_list = [TEST_DOCS[i] for i in sorted(TEST_DOCS)]
for fold in CFG["FOLDS_TO_RUN"]:
    if fold in hist: print(f"fold {fold} already done, skipping"); continue
    dev_ids = set(FOLDS[fold]); tr = [DOCS[i] for i in ALL_IDS if i not in dev_ids]; dv = [DOCS[i] for i in FOLDS[fold]]
    model, h = train_fold(tr, dv, fold)
    oof.update(predict_probs(model, dv)); test_probs.append(predict_probs(model, test_docs_list)); hist[fold] = h
    pickle.dump(dict(oof=oof, test_probs=test_probs, hist=hist, tags=TAGS, cfg=CFG), open(done_path, "wb"))
    del model; torch.cuda.empty_cache()

# ---- results ----
docs_oof = [DOCS[i] for i in ALL_IDS if i in oof]
spans = decode(oof, docs_oof)
f1, p, r, per = strict_f1({d.id: d.gold for d in docs_oof}, spans)
print(f"\n=== OUT-OF-FOLD (each doc predicted by a model that never saw it): {len(docs_oof)} docs ===")
print(f"strict macro-F1 {f1:.4f}   precision {p:.4f}   recall {r:.4f}")
for l in LABELS: print(f"   {l:10s} F1 {per[l]['f1']:.3f}  P {per[l]['p']:.3f}  R {per[l]['r']:.3f}  n={per[l]['support']}")

# per-label bias tuned on the out-of-fold predictions (raises macro-F1 directly; coarse grid to avoid overfitting)
GRID = [-1.0, -0.5, 0.0, 0.5, 1.0, 1.5, 2.0]
def score_bias(b): return strict_f1({d.id: d.gold for d in docs_oof}, decode(oof, docs_oof, b))[0]
bias = {l: 0.0 for l in LABELS}; best = f1
for _ in range(2):
    for l in LABELS:
        for v in GRID:
            trial = dict(bias, **{l: v}); s = score_bias(trial)
            if s > best + 1e-6: best, bias = s, trial
print(f"\nwith per-label bias tuned on the folds: macro-F1 {best:.4f} (was {f1:.4f});  bias = { {k: v for k, v in bias.items() if v} }")

# ---- test predictions: average the fold models ----
def write_submission(name, b):
    n = len(test_probs)
    avg = {d.id: sum(tp[d.id] for tp in test_probs) / n for d in test_docs_list}
    sp_ = decode(avg, test_docs_list, b); pred = {}
    for d in test_docs_list:
        pred[d.id] = [[int(s), int(e), l] for s, e, l in sp_[d.id] if 0 <= s < e <= len(d.text) and l in LABELS]
    assert set(pred) == set(test_texts), "test ids mismatch"
    out = RUN_DIR / name; out.mkdir(exist_ok=True)
    json.dump(pred, open(out / "prediction.json", "w"))
    with zipfile.ZipFile(RUN_DIR / f"submission_{name}.zip", "w", zipfile.ZIP_DEFLATED) as z: z.write(out / "prediction.json", "prediction.json")
    cnt = collections.Counter(x[2] for v in pred.values() for x in v)
    print(f"{name}: {sum(len(v) for v in pred.values())} entities in {len(pred)} docs  {dict(cnt)}")
    return avg
if test_probs:
    print(f"\nTest predictions from {len(test_probs)} fold model(s):")
    write_submission("raw", None); write_submission("tuned", bias)
    print(f"\nUpload this file to Codabench: {RUN_DIR}/submission_tuned.zip  (or submission_raw.zip)")
    print("Also keep results.pkl - the later ensemble step uses it.")
```

```python
# ---- Error analysis on the out-of-fold predictions (runs in seconds, no training) ----
def diagnose(oof_probs, bias=None, show=4):
    docs = [DOCS[i] for i in ALL_IDS if i in oof_probs]
    pred = decode(oof_probs, docs, bias)
    rows = {l: collections.Counter() for l in LABELS}; ex = {l: [] for l in LABELS}
    for d in docs:
        P = pred[d.id]; G = d.gold
        gset = {(s, e, l) for s, e, l in G}; gspan = {(s, e): l for s, e, l in G}
        for s, e, l in G:
            c = rows[l]; c["gold"] += 1
            if (s, e, l) in set(P): c["exact"] += 1; continue
            if any(ps == s and pe == e for ps, pe, pl in P): c["right span, wrong label"] += 1; continue
            ov = [(ps, pe, pl) for ps, pe, pl in P if ps < e and pe > s]
            if any(pl == l for _, _, pl in ov):
                c["right label, wrong boundary"] += 1
                if len(ex[l]) < show:
                    m = next((ps, pe) for ps, pe, pl in ov if pl == l); ex[l].append((d.text[s:e], d.text[m[0]:m[1]]))
            elif ov: c["overlaps a different label"] += 1
            else: c["missed completely"] += 1
        for s, e, l in P:
            if not any(ps < e and pe > s for ps, pe, pl in G): rows[l]["spurious (no gold nearby)"] += 1
    keys = ["gold", "exact", "right label, wrong boundary", "right span, wrong label", "overlaps a different label", "missed completely", "spurious (no gold nearby)"]
    print(f"{'label':10s}" + "".join(f"{k.split(',')[0][:13]:>15s}" for k in keys))
    for l in LABELS: print(f"{l:10s}" + "".join(f"{rows[l][k]:15d}" for k in keys))
    tot = collections.Counter(); [tot.update(rows[l]) for l in LABELS]
    g = tot["gold"]
    print("\nOf all gold entities: " + ", ".join(f"{100*tot[k]/g:.0f}% {k}" for k in keys[1:6]))
    for l in LABELS:
        if ex[l]: print(f"\n{l} boundary examples (gold  ->  predicted):"); [print(f"   {a!r}  ->  {b!r}") for a, b in ex[l]]
diagnose(oof)
```
